Problem Statement:

A backend engineer is building a lightweight semantic search microservice for an internal ML knowledge base containing short factual sentences about ML algorithms. The engineer needs to implement three components:

A cosine_similarity function using NumPy.
A search_notes function that takes a plain-text query, computes cosine similarity between the query embedding and every stored knowledge-base embedding, and returns the top-k most similar entries with their scores.
A chunk_by_paragraph function that splits a raw document string into chunks, keeping only paragraphs that meet a minimum word count.
The engineer does NOT have access to the OpenAI API in this environment. A stub get_embeddings(text) function is provided below that simulates embeddings using deterministic NumPy random vectors seeded by the text's length. The solution must work with this stub.

Constraints & Requirements:

Use only numpy (imported as np). No other third-party libraries.
cosine_similarity must accept two plain Python lists and return a Python float.
search_notes must return a list of (score, document) tuples sorted descending by score, limited to top_k results.
chunk_by_paragraph must split on "\n", strip whitespace from each resulting line, and keep only lines with at least min_words words (default 20).
The stub get_embeddings below is provided and must not be modified.

```python
import numpy as np

# --- PROVIDED STUB (do not modify) ---
def get_embeddings(text):
    """Stub: deterministic 8-dimensional fake embedding for testing."""
    rng = np.random.default_rng(len(text) % 1000)
    vec = rng.standard_normal(8)
    return vec.tolist()

# --- Part 1: cosine_similarity ---
def cosine_similarity(vector1, vector2):
    """
    Compute cosine similarity between two vectors.
    Accepts Python lists and returns a Python float.
    """
    v1 = np.array(vector1, dtype=float)
    v2 = np.array(vector2, dtype=float)

    norm1 = np.linalg.norm(v1)
    norm2 = np.linalg.norm(v2)

    if norm1 == 0 or norm2 == 0:
        return 0.0

    similarity = np.dot(v1, v2) / (norm1 * norm2)
    return float(similarity)

# --- Part 2: Pre-compute knowledge-base embeddings ---
knowledge_base = [
    "Linear regression predicts a continuous number using a straight line.",
    "Decision tree splits the data based on feature values.",
    "Random forest combines many decision trees and lets them vote.",
    "Logistic regression predicts probability values for classification.",
    "Overfitting happens when a model memorizes the data.",
]

kb_embeddings = {
    doc: get_embeddings(doc)
    for doc in knowledge_base
}

def search_notes(query, top_k=3):
    """
    Returns a list of (score, document) tuples sorted
    by descending cosine similarity.
    """
    query_embedding = get_embeddings(query)

    results = []
    for doc, embedding in kb_embeddings.items():
        score = cosine_similarity(query_embedding, embedding)
        results.append((score, doc))

    results.sort(key=lambda x: x[0], reverse=True)
    return results[:top_k]

# --- Part 3: chunk_by_paragraph ---
def chunk_by_paragraph(document, min_words=20):
    """
    Split document on newline, strip whitespace,
    and keep only paragraphs with at least min_words words.
    """
    chunks = []

    for line in document.split("\n"):
        paragraph = line.strip()
        if paragraph and len(paragraph.split()) >= min_words:
            chunks.append(paragraph)

    return chunks

# --- Test ---
if __name__ == "__main__":
    sample_doc = """Short line.

Linear regression predicts a continuous number using a straight line so it is useful for forecasting tasks in many real-world scenarios and business applications.

Decision tree splits the data based on feature values and produces an interpretable model structure suitable for classification or regression problems.

Hi."""

    print(chunk_by_paragraph(sample_doc, min_words=20))
    print(search_notes("What algorithm uses multiple trees?", top_k=3))
```
